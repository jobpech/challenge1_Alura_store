### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda1 = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)
```

#1. Análisis de facturación



En este primer análisis, debes calcular el ingreso total de cada tienda. Sumando los valores de la columna Precio de cada conjunto de datos de la tienda para estimar los ingresos.

```python
# Añadimos columna 'Ingresos' para cada tienda
tienda1['Ingresos'] = tienda1['Precio']
tienda2['Ingresos'] = tienda2['Precio']
tienda3['Ingresos'] = tienda3['Precio']
tienda4['Ingresos'] = tienda4['Precio']
```

```python
ingresos_tienda1 = tienda1['Ingresos'].sum()
ingresos_tienda2 = tienda2['Ingresos'].sum()
ingresos_tienda3 = tienda3['Ingresos'].sum()
ingresos_tienda4 = tienda4['Ingresos'].sum()

print("Tienda totales de la tienda 1: ", ingresos_tienda1,
      "\nTienda totales de la tienda 2: ", ingresos_tienda2,
      "\nTienda totales de la tienda 3: ", ingresos_tienda3,
      "\nTienda totales de la tienda 4: ", ingresos_tienda4
      )
print("La tienda 4 tiene menor ingresos con un total de: ", ingresos_tienda4)


```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Crear un DataFrame con los datos
df_ingresos = pd.DataFrame({
    'Tienda': ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4'],
    'Ingresos': [ingresos_tienda1, ingresos_tienda2, ingresos_tienda3, ingresos_tienda4]
})

# Paleta de colores personalizada
colores = ['royalblue', 'orange', 'forestgreen', 'orchid']

# Crear la gráfica con seaborn
plt.figure(figsize=(10, 3))
sns.barplot(data=df_ingresos, x='Tienda', y='Ingresos', palette=colores, hue='Tienda', legend=False)

# Personalización de estilo
plt.title('Ingresos totales por tiendas', loc='left', fontsize=18)
plt.ylabel('Ingresos', fontsize=14)
plt.xlabel('')
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
# Guardar la gráfica como imagen
plt.tight_layout()
plt.savefig('ingresos_totales_por_tienda.jpg')  # o .jpg, .svg, .pdf
plt.show()
```

# 2. Ventas por categoría

En este debes calcular la cantidad de productos vendidos por categoría en cada tienda. La idea es agrupar los datos por categoría y contar el número de ventas de cada tipo, mostrando las categorías más populares de cada tienda.

```python
tienda1['Tienda'] = 'Tienda 1'
tienda2['Tienda'] = 'Tienda 2'
tienda3['Tienda'] = 'Tienda 3'
tienda4['Tienda'] = 'Tienda 4'
```

```python
df_total = pd.concat([tienda1, tienda2, tienda3, tienda4], ignore_index=True)
df_total
```

```python
ventas_categoria = df_total.groupby(['Tienda', 'Categoría del Producto'])['Producto'].count().reset_index()
ventas_categoria.rename(columns={'Producto': 'Cantidad Vendida'}, inplace=True)
ventas_categoria
```

```python
# Mostrar las 3 categorías más vendidas por cada tienda
top_categorias_por_tienda = ventas_categoria.sort_values(['Tienda', 'Cantidad Vendida'], ascending=[True, False])


for tienda in top_categorias_por_tienda['Tienda'].unique():
    print(f"\n {tienda} - Categorías más vendidas:")
    print(top_categorias_por_tienda[top_categorias_por_tienda['Tienda'] == tienda].head(3))
```

```python
ventas_categoria = (
    df_total.groupby(['Tienda', 'Categoría del Producto'])['Producto']
    .count()
    .reset_index()
    .rename(columns={'Producto': 'Cantidad Vendida'})
)
```

```python
# Obtener top 3 por tienda
top3_por_tienda = ventas_categoria.sort_values(['Tienda', 'Cantidad Vendida'], ascending=[True, False]) \
    .groupby('Tienda') \
    .head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Obtener el máximo valor global para unificar escala
maximo_global = top3_por_tienda['Cantidad Vendida'].max()

# Crear figura con 4 subplots (2x2)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

# Obtener tiendas únicas
tiendas = top3_por_tienda['Tienda'].unique()

# Crear una gráfica por tienda
for i, tienda in enumerate(tiendas):
    datos_tienda = top3_por_tienda[top3_por_tienda['Tienda'] == tienda].head(3)

    sns.barplot(
        data=datos_tienda,
        x='Categoría del Producto',
        y='Cantidad Vendida',
        ax=axes[i],
        palette='Set2',
        hue='Categoría del Producto',
        legend=False
    )

    axes[i].set_title(tienda, fontsize=13)
    axes[i].set_ylabel('Cantidad Vendida')
    axes[i].set_xlabel('Categoría')
    axes[i].tick_params(axis='x', rotation=45)
    axes[i].set_ylim(0, maximo_global + 1)  # <- misma escala para todas

# Eliminar ejes vacíos si hay menos de 4 tiendas
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Título global
plt.suptitle('Top 3 Categorías Más Vendidas por Tienda', fontsize=16)

# Ajuste de márgenes y guardado
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('categorias_mas_vendidas_por_tienda.jpg', dpi=300)
plt.show()
```

# 3. Calificación promedio de la tienda


En este paso, debes calcular las calificaciones promedio de los clientes para cada tienda. El objetivo es conocer la satisfacción del cliente con los productos vendidos.

```python
calificaciones_promedio = df_total.groupby('Tienda')['Calificación'].mean().reset_index()
calificaciones_promedio.rename(columns={'Calificación': 'Calificación Promedio'}, inplace=True)

print(calificaciones_promedio)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
sns.barplot(
    data=calificaciones_promedio,
    x='Tienda',
    y='Calificación Promedio',
    palette='Set2',
    hue='Tienda',
    legend=False
)
plt.title('Calificación Promedio por Tienda')
plt.ylim(0, 5)
plt.ylabel('Calificación Promedio (de 5)')
plt.xlabel('Tienda')
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
# Guardar la gráfica como imagen
plt.savefig('calificacion_promedio_por_tienda.jpg')  # o .jpg, .svg, .pdf
plt.show()
```

# 4. Productos más y menos vendidos

En este paso, debes identificar qué productos fueron los más vendidos y los menos vendidos en cada tienda. Visualiza los resultados para que quede claro qué productos destacaron en ventas en cada tienda.

```python
ventas_por_producto = df_total.groupby(['Tienda', 'Producto'])['Producto'].count().reset_index(name='Cantidad Vendida')
```

```python
top_productos = []
bottom_productos = []

for tienda in ventas_por_producto['Tienda'].unique():
    datos_tienda = ventas_por_producto[ventas_por_producto['Tienda'] == tienda]
    top = datos_tienda.sort_values('Cantidad Vendida', ascending=False).head(3)
    bottom = datos_tienda.sort_values('Cantidad Vendida', ascending=True).head(3)

    top_productos.append(top)
    bottom_productos.append(bottom)

top_df = pd.concat(top_productos)
bottom_df = pd.concat(bottom_productos)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Calcular el valor máximo global de "Cantidad Vendida" para usar en xlim()
maximo_global = top_df['Cantidad Vendida'].max()

# Obtener lista de tiendas únicas
tiendas = top_df['Tienda'].unique()

# Crear figura y ejes para subplots (2 filas, 2 columnas)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()  # Convierte a lista para facilitar el acceso por índice

# Gráficas por tienda
for i, tienda in enumerate(tiendas):
    datos = top_df[top_df['Tienda'] == tienda].sort_values('Cantidad Vendida', ascending=False)

    sns.barplot(
        data=datos,
        x='Cantidad Vendida',
        y='Producto',
        palette='Set2',
        ax=axes[i],
        hue='Producto',
        legend=False
    )

    axes[i].set_title(f'Top 3 Productos Más Vendidos - {tienda}', fontsize=12)
    axes[i].set_xlim(0, maximo_global + 1)  # Misma escala
    axes[i].set_xlabel('Cantidad Vendida')
    axes[i].set_ylabel('Producto')

# Ajustar espacio entre subplots
plt.tight_layout()

# Guardar la figura completa
plt.savefig('top3_productos_mas_vendidos_por_tienda.jpg', dpi=300)

# Mostrar en pantalla
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Estilo general
sns.set(style='whitegrid')

# Obtener el máximo valor global de ventas para fijar escala común
maximo_global = bottom_df['Cantidad Vendida'].max()

# Lista de tiendas
tiendas = bottom_df['Tienda'].unique()

# Crear figura con subplots 2x2
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
axes = axes.flatten()  # Aplanar para iterar más fácil

# Colores por consistencia
palette = 'Set2'

# Graficar cada tienda en su subplot
for i, tienda in enumerate(tiendas):
    datos = bottom_df[bottom_df['Tienda'] == tienda].sort_values('Cantidad Vendida', ascending=True)

    sns.barplot(
        data=datos,
        x='Cantidad Vendida',
        y='Producto',
        palette=palette,
        ax=axes[i],
        hue = 'Producto',
        legend= False
    )

    axes[i].set_title(f'Top 3 Productos Menos Vendidos - {tienda}', fontsize=13)
    axes[i].set_xlabel('Cantidad Vendida')
    axes[i].set_ylabel('Producto')
    axes[i].set_xlim(0, maximo_global + 1)  # Misma escala para todos

# Quitar paneles vacíos si hay menos de 4 tiendas (por precaución)
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.savefig('productos_menos_vendidos_4_tiendas.jpg', dpi=300)
plt.show()
```

# 5. Envío promedio por tienda

En este paso, debes calcular el costo de envío promedio para cada tienda. El objetivo es comprender cuánto se gasta, en promedio, en el envío de cada tienda.

```python
costo_envio_promedio = df_total.groupby('Tienda')['Costo de envío'].mean().reset_index()
costo_envio_promedio.rename(columns={'Costo de envío': 'Costo de Envío Promedio'}, inplace=True)

print(costo_envio_promedio)
```

```python
import matplotlib.pyplot as plt

# Datos
etiquetas = costo_envio_promedio['Tienda']
valores = costo_envio_promedio['Costo de Envío Promedio']
colores = ['royalblue', 'orange', 'forestgreen', 'orchid']

# Gráfico de pastel
plt.figure(figsize=(7, 7))
plt.pie(
    valores,
    labels=etiquetas,
    autopct='%1.1f%%',         # Muestra porcentaje con 1 decimal
    startangle=140,            # Ángulo de inicio
    colors=colores,
    wedgeprops={'edgecolor': 'white'}  # Bordes blancos entre sectores
)
plt.title('Distribución del Costo de Envío Promedio por Tienda', fontsize=14)
plt.tight_layout()
# Guardar la gráfica como imagen
plt.savefig('costo_envio_promedio.jpg')  # o .jpg, .svg, .pdf
plt.show()
```
